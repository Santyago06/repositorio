```python
# a)
import sympy as sp

# Definir variable
x = sp.symbols('x', real=True)

# Producto interno en [-1,1]
def ip(f, g):
    return sp.integrate(f*g, (x, -1, 1))

# Base de monomios hasta cierto grado
n = 4
basis = [x**k for k in range(n+1)]

# Calcular matriz de Gram
G = sp.Matrix([[ip(f,g) for g in basis] for f in basis])

print("Base usada:", basis)
print("\nMatriz de Gram (productos internos):")
sp.pprint(G)

# Verificar si hay términos no nulos fuera de la diagonal
non_orthogonal = [(i,j,G[i,j]) for i in range(len(basis)) for j in range(i+1,len(basis)) if G[i,j] != 0]

print("\nPares no ortogonales:")
for i,j,val in non_orthogonal:
    print(f"<x^{i}, x^{j}> = {val}")
```

```python
#d.I)
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np

# Variable simbólica
x = sp.symbols('x')

# Definir la función
h = sp.sin(3*x)*(1 - x**2)

#Expansión en base de monomios
N = 10  # grado máximo
coeffs_monomios = []
monomios_exp = 0
for n in range(N+1):
    phi = x**n
    num = sp.integrate(h*phi, (x, -1, 1))
    den = sp.integrate(phi*phi, (x, -1, 1))
    cn1 = sp.simplify(num/den)
    coeffs_monomios.append(cn1)
    monomios_exp += cn1*phi

print("Coeficientes en la base de monomios:")
print(coeffs_monomios)
print("\nExpansión en monomios:")
print(sp.simplify(monomios_exp))

# Expansión en base de polinomios de Legendre
coeffs_legendre = []
legendre_exp = 0
for n in range(N+1):
    Pn = sp.legendre(n, x)
    num = sp.integrate(h*Pn, (x, -1, 1))
    den = sp.integrate(Pn*Pn, (x, -1, 1))
    cn3 = sp.simplify(num/den)
    coeffs_legendre.append(cn3)
    legendre_exp += cn3*Pn

print("\nCoeficientes en la base de Legendre:")
print(coeffs_legendre)
print("\nExpansión en polinomios de Legendre:")
print(sp.simplify(legendre_exp))

# Gráficas
f_h = sp.lambdify(x, h, 'numpy')
f_monomios = sp.lambdify(x, monomios_exp, 'numpy')
f_legendre = sp.lambdify(x, legendre_exp, 'numpy')

xx = np.linspace(-1, 1, 400)
plt.plot(xx, f_h(xx), 'k', label='h(x)')
plt.plot(xx, f_monomios(xx), 'r--', label='Monomios')
plt.plot(xx, f_legendre(xx), 'b-.', label='Legendre')
plt.legend()
plt.title("Comparación de expansiones")
plt.grid(True)
plt.show()
```

```python
# d.II)
import mpmath as mp

# Variable
x = sp.symbols('x')

# Producto interno
def p_i(f,g):
  m = sp.simplify(f*g*sp.sqrt(1-x**2))
  f_num = sp.lambdify(x,m,"mpmath")
  val = mp.quad(f_num, [-1,1])
  return val
# Función h(x)
h = sp.sin(3*x)*(1-x**2)

# Expansión de monomios
N = 10
coeffs_monomios = []
monomios_exp = 0
for n in range(N+1):
    phi = x**n
    cn1 = sp.simplify(p_i(h,phi)/p_i(phi,phi))
    coeffs_monomios.append(cn1)
    monomios_exp += cn1*phi

print("Coeficientes en la base de monomios:")
print(coeffs_monomios)
print("\nExpansión en monomios:")
print(sp.simplify(monomios_exp))


# Expansión de polinomios de chebyshev
coeffs_chebU = []
chebU_exp = 0
for n in range (N+1):
  U = sp.chebyshevu(n, x)
  cn2 = sp.simplify(p_i(h,U)/p_i(U,U))
  coeffs_chebU.append(cn2)
  chebU_exp += cn2*U

print("\nCoeficientes en la base de Chebyshev:")
print(coeffs_chebU)
print("\nExpansión en Chebyshev:")
print(sp.simplify(chebU_exp))

# Gráficas
f_h = sp.lambdify(x, h, 'numpy')
f_monomios = sp.lambdify(x, monomios_exp, 'numpy')
f_chebU = sp.lambdify(x, chebU_exp, 'numpy')

xx = np.linspace(-1, 1, 400)
plt.plot(xx, f_h(xx), 'k', label='h(x)')
plt.plot(xx, f_monomios(xx), 'r--', label='Monomios')
plt.plot(xx, f_chebU(xx), 'b-.', label='Chebyshev')
plt.legend()
plt.title("Comparación de expansiones")
plt.grid(True)
plt.show()
```
