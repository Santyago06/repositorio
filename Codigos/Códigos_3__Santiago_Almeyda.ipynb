```python
import sympy as sp

# PUNTO 1.A

# Variables
x, y = sp.symbols('x y')
a0, a1, a2, b0, b1, b2 = sp.symbols('a0 a1 a2 b0 b1 b2')

# P2(x) y G2(y)
h = (a0 + a1*x + a2*x**2)*(b0 + b1*y + b2*y**2)

# bases monomiales
base_x_monom = [1, x, x**2]
base_y_monom = [1, y, y**2]

# C^ij
coeffs_monom = []
for i in range(3):
    for j in range(3):
        coeffs_monom.append(sp.expand(h).coeff(x, i).coeff(y, j))

coeffs_monom = sp.Matrix(coeffs_monom)

print("Coeficientes en base monomial (columna 9x1):\n")
sp.pprint(coeffs_monom)
```

```python
# PUNTO 1.B

# P2(x) ⊗ G2 (x)
p = 3 + x + x**2
g = 1 + y
T = p*g
print("T(x,y) =", sp.expand(T))

# C^ij
coeffs_monom = []
for i in range(3):
    for j in range(3):
        coeffs_monom.append(sp.expand(T).coeff(x, i).coeff(y, j))

coeffs_monom = sp.Matrix(coeffs_monom)

print("\nCoeficientes en base monomial (columna 9x1):\n")
sp.pprint(coeffs_monom)
```

```python
# PUNTO 1.C

def p_i(f,k):
  return sp.integrate(f*k, (x,-1,1))

# Expansión de Legendre de P2(x)
N = 2
coeffs_legendre = []
legendre_exp = 0
for n in range(N+1):
    Pn = sp.legendre(n, x)
    c = sp.simplify(p_i(p,Pn)/p_i(Pn,Pn))
    coeffs_legendre.append(c)
    legendre_exp += c*Pn

print("\nCoeficientes en la base de Legendre:")
print(coeffs_legendre)
print("\nExpansión en polinomios de Legendre:")
print(sp.simplify(legendre_exp))
```

```python
# PUNTO 1.D

P0, P1, P2 = sp.legendre(0, x), sp.legendre(1, x), sp.legendre(2, x)
Q0, Q1, Q2 = sp.legendre(0, y), sp.legendre(1, y), sp.legendre(2, y)

base_x_leg = [P0, P1, P2]
base_y_leg = [Q0, Q1, Q2]

coeffs_leg = []
for i in range(3):
    for j in range(3):
        Bij = base_x_leg[i]*base_y_leg[j]
        num = sp.integrate(sp.integrate(T*Bij, (x,-1,1)), (y,-1,1))
        den = sp.integrate(sp.integrate(Bij*Bij, (x,-1,1)), (y,-1,1))
        coeffs_leg.append(sp.simplify(num/den))

coeffs_leg = sp.Matrix(coeffs_leg)

print("\nCoeficientes en base de Legendre (columna 9x1):\n")
sp.pprint(coeffs_leg)
```
